```python
import random
import string
Action = input("Do you want to generate a password or check a passwrod? (select Generate or Check) ")
if Action == "Generate":
  print("Welcome to Password Generator")
  print("Your password should include letter, symbols and numbers")
  lenght = int(input("Please enter the lenght of your password: "))
  characters = string.ascii_letters + string.digits + string.punctuation
  password = ""
  for i in range(lenght):
    random_characters = random.choice(characters)
    password = password + random_characters
  print(f"Your generated password is {password}")
elif Action == "Check":
 import string

password = input("Please enter your password: ")

has_lower = False
has_upper = False
has_digit = False
has_symbol = False

for i in password:
    if i in string.ascii_lowercase:
        has_lower = True
    elif i in string.ascii_uppercase:
        has_upper = True
    elif i in string.digits:
        has_digit = True
    elif i in string.punctuation:
        has_symbol = True

point = 0

if has_lower:
    point += 1
if has_upper:
    point += 1
if has_digit:
    point += 1
if has_symbol:
    point += 1

if point <= 2:
    print("Weak password!")
elif point == 3:
    print("Medium password!")
else:
    print("Powerful password!")
```
